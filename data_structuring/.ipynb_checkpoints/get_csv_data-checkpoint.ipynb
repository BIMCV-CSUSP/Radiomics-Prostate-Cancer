# Proposito del notebook

Este notebook recopila y estructura los datos del conjunto de datos del reto PI-CAI, generando un archivo CSV (`data.csv`) que centraliza toda la información del dataset. El proceso incluye: 

1. Indexar las rutas a las secuencias de imágenes (T2W, ADC, HBV) para cada paciente y estudio
2. Integrar datos clínicos (fecha de MRI, edad, PSA, densidad de PSA, volumen prostático, puntuación ISUP, clasificación csPCa)
3. Localizar e incorporar las rutas a las delineaciones de las lesiones cancerosas, identificando su origen (experto humano o algoritmo de IA)
4. Extraer las dimensiones cada imagen (altura, ancho, profundidad)
5. Identificar valores de la puntuación ISUP a partir de las lesiones
6. Añadir rutas a las segmentaciones anatómicas completas (glándula prostática) y zonales (zona periférica/transicional).

El CSV resultante resultante sirve como base de datos relacional para los análisis posteriores, facilitando el acceso organizado a las 1500 muestras del conjunto de datos con sus correspondientes imágenes, etiquetas y datos clínicos asociados.

# Image paths

El siguiente código construye un diccionario con las rutas a las imágenes médicas `.mha` de cada paciente y cada estudio. El directorio `../../data/images` contiene un subdirectorio por paciente, y dentro de cada uno puede haber uno o más estudios, reflejados en los nombres de los archivos, que combinan el identificador del paciente (`patient_id`) y del estudio (`study_id`), seguidos del tipo de imagen (`t2w`, `adc` o `hbv`).

El objetivo del siguiente código es recorrer todos los archivos `.mha`, extraer la información del paciente y el estudio a partir del nombre del archivo, y construir un diccionario `data_dict` donde:

- Cada clave es una tupla `(patient_id, study_id)`.
- Cada valor es un diccionario con:
  - Los identificadores (`patient_id` y `study_id`).
  - Las rutas relativas a cada tipo de imagen (`t2w_path`, `adc_path`, `hbv_path`), si están disponibles.

Este diccionario se convierte en un `DataFrame` que relaciona `(patient_id, study_id)` con las rutas a sus distintas secuencias de imagen.

```python
image_dir = "../../data/images" # Ruta absoluta al directorio de imágenes
relative_base = "data/images" # Prefijo para las rutas relativas

# Crear diccionario vacío para almacenar información de imágenes
data_dict = {}

# Recorrer recursivamente el directorio de imágenes
for subdir, _, files in os.walk(image_dir):
    for file in files:
        # Filtrar archivos .mha (formato de imagen médica)
        if file.endswith(".mha"):
            # Construir rutas absolutas y relativas
            abs_path = os.path.join(subdir, file)
            relative_path = os.path.relpath(abs_path, start=image_dir)
            
            # Extraer IDs de paciente y estudio del nombre del archivo
            # Ejemplo: "10129_100130_t2w.mha" -> patient_id=10129, study_id=100130
            file_name = os.path.splitext(file)[0]
            patient_id, study_id = file_name.split('_')[0], file_name.split('_')[1]
            
            # Crear clave compuesta para el diccionario
            key = (patient_id, study_id)
            
            # Inicializar registro si no existe para este paciente/estudio
            if key not in data_dict:
                data_dict[key] = {
                    "patient_id": patient_id,
                    "study_id": study_id,
                    "t2w_path": None,
                    "adc_path": None,
                    "hbv_path": None
                }
            
            # Asignar ruta según el tipo de secuencia identificada en el nombre
            if "t2w" in file:
                data_dict[key]["t2w_path"] = os.path.join(relative_base, relative_path)
            elif "adc" in file:
                data_dict[key]["adc_path"] = os.path.join(relative_base, relative_path)
            elif "hbv" in file:
                data_dict[key]["hbv_path"] = os.path.join(relative_base, relative_path)


# Crear DataFrame con la información recopilada
data = list(data_dict.values())
df = pd.DataFrame(data)
```

# Clinical_data

El siguiente código integra los datos clínicos de los pacientes con el `DataFrame` previamente creado que contiene las rutas de las imágenes. Para ello:

1. Carga el archivo `marksheet.csv` que contiene información clínica detallada de los pacientes.  
2. Convierte los identificadores de paciente y estudio a formato de texto para asegurar una correcta correspondencia durante la fusión.  
3. Realiza una operación de `merge` (unión) entre el `DataFrame` de rutas de imágenes y el `DataFrame` de información clínica, usando como claves los campos `patient_id` y `study_id`.  
4. Convierte la variable categórica `case_csPCa` (cáncer de próstata clínicamente significativo) de formato texto (`'YES'`/`'NO'`) a binario (`1`/`0`).  
5. Reorganiza las columnas del `DataFrame` para una mejor lectura, colocando los datos clínicos entre los identificadores de paciente/estudio y las rutas de las imágenes.

El `DataFrame` resultante (`df_2`) ahora contiene tanto las rutas a las imágenes como los datos clínicos relevantes: fecha de la resonancia, edad del paciente, nivel de PSA, densidad de PSA, volumen prostático, puntuación ISUP y presencia de cáncer clínicamente significativo.

```python
# Cargar los datos clínicos desde el archivo marksheet.csv
clinical_df = pd.read_csv("../../data/labels/clinical_information/marksheet.csv")

# Convertir los IDs a formato string para asegurar la compatibilidad en la unión
df['patient_id'] = df['patient_id'].astype(str)
df['study_id'] = df['study_id'].astype(str)
clinical_df['patient_id'] = clinical_df['patient_id'].astype(str)
clinical_df['study_id'] = clinical_df['study_id'].astype(str)

# Fusionar el DataFrame de imágenes con los datos clínicos
df_merged = pd.merge(
    df,
    clinical_df[['patient_id', 'study_id', 'mri_date', 'patient_age', 'psa', 'psad',
                 'prostate_volume', 'case_ISUP', 'case_csPCa']],
    on=['patient_id', 'study_id'],
    how='left'
)

# Convertir la variable de cáncer clínicamente significativo a formato binario
df_merged['case_csPCa'] = df_merged['case_csPCa'].map({'NO': 0, 'YES': 1})

# Reordenar las columnas para mejor legibilidad
# Colocamos los datos clínicos después de los identificadores pero antes de las rutas
cols_to_insert = ['mri_date', 'patient_age', 'psa', 'psad', 'prostate_volume',
                  'case_ISUP', 'case_csPCa']
existing_cols = df.columns.tolist()
new_col_order = existing_cols[:2] + cols_to_insert + existing_cols[2:]
df_2 = df_merged[new_col_order]

# DataFrame resultante
df_2
```

# csPCa_lesion_delineations

Esta sección del código localiza las segmentaciones de las lesiones de cáncer de próstata clínicamente significativo (csPCa) para cada paciente y estudio. Las delineaciones pueden provenir de dos fuentes: expertos humanos o un algoritmo de IA (`Bosma22a`). 

El código busca los archivos de segmentación correspondientes (en formato NIfTI, `.nii.gz`) en ambas ubicaciones, priorizando las delineaciones realizadas por expertos humanos. Además de guardar la ruta a la delineación, se crean dos columnas (`human_delineation` y `AI_delineation`) que especifican el origen de la segmentación (`1` para presente, `0` para ausente).

```python
# Definir rutas a los directorios que contienen las delineaciones de lesiones csPCa
human_dir = "../../data/labels/csPCa_lesion_delineations/human_expert/resampled" # Segmentaciones por expertos
ai_dir = "../../data/labels/csPCa_lesion_delineations/AI/Bosma22a" # Segmentaciones por IA
```

```python
def get_delineation_path(row):
    # Construir nombre de archivo basado en patient_id y study_id
    file_name = f"{row['patient_id']}_{row['study_id']}.nii.gz"
    
    # Construir rutas completas a posibles ubicaciones del archivo
    human_path = os.path.join(human_dir, file_name)
    ai_path = os.path.join(ai_dir, file_name)
    
    # Verificar existencia de archivos, priorizando delineaciones humanas
    if os.path.exists(human_path):
        # Si existe delineación humana, devolver: (ruta_relativa, humano=1, IA=0)
        return human_path.replace("../../", ""), 1, 0 
    elif os.path.exists(ai_path):
        # Si existe delineación por IA, devolver: (ruta_relativa, humano=0, IA=1)
        return ai_path.replace("../../", ""), 0, 1
    else:
        # Si no existe ninguna delineación, devolver: (NaN, humano=0, IA=0)
        return np.nan, 0, 0 
```

```python
# Aplicar la función a cada fila del DataFrame y asignar los resultados a nuevas columnas
df_2[['csPCa_lesion_delineation_path', 'human_delineation', 'AI_delineation']] = df_2.apply(
    lambda row: pd.Series(get_delineation_path(row)), axis=1
)
```

# Shape and ISUP (again)

Esta sección del código extrae las dimensiones de las imágenes T2W y los valores de clasificación ISUP directamente de las máscaras de segmentación de lesiones. Utilizando la biblioteca MONAI, el código carga cada imagen T2W junto con su correspondiente máscara de segmentación de lesión cancerosa, y registra las dimensiones espaciales de las imágenes (altura, ancho, profundidad). Además, analiza las máscaras de segmentación para determinar los valores de ISUP codificados en ellas (valores distintos de cero, que normalmente representa tejido sano). En casos donde se detectan múltiples valores de ISUP en la misma máscara (indicando posibles múltiples lesiones de diferente grado), se imprime esta información para su revisión y se selecciona el valor más alto (correspondiente a la lesión de mayor grado, representada por el valor numérico más grande). Estos valores de dimensión y grados de lesión extraídos directamente de las segmentaciones se almacenan en el DataFrame.

```python
# Usar el DataFrame anterior como base
df_3 = df_2

# Inicializar nuevas columnas para dimensiones de imágenes y valores ISUP de lesiones
df_3[['height', 'width', 'depth']] = np.nan
df_3['lesion_value'] = np.nan

# Configurar pipeline de transformaciones MONAI para cargar y procesar imágenes
transforms = monai.transforms.Compose([
    # Cargar imágenes T2 y máscaras de lesiones, asegurando formato de canal adecuado
    monai.transforms.LoadImaged(keys=['t2', 'label'], image_only=False, ensure_channel_first=True),
    # Normalizar intensidades de imágenes T2 al rango [0,1]
    monai.transforms.ScaleIntensityd(keys=['t2'], minv=0.0, maxv=1.0, allow_missing_keys=True)
])

# Procesar cada fila del DataFrame
for idx, row in df_3.iterrows():
    patient_id = row['patient_id']
    study_id = row['study_id']

    # Construir rutas completas a la imagen T2 y su máscara de segmentación de lesión
    t2_path = f"../../{row['t2w_path']}" if pd.notna(row['t2w_path']) else None
    label_path = f"../../{row['csPCa_lesion_delineation_path']}" if pd.notna(row['csPCa_lesion_delineation_path']) else None

    # Verificar que ambos archivos existan
    if t2_path and os.path.exists(t2_path) and label_path and os.path.exists(label_path):
        # Crear diccionario con las rutas para el procesamiento MONAI
        image_dict = {
            't2': t2_path,
            'label': label_path
        }
        
        # Aplicar transformaciones (cargar y normalizar)
        processed_images = transforms(image_dict)

        # Extraer dimensiones espaciales de la imagen T2
        t2_shape = processed_images['t2'].shape[1:] 
        df_3.loc[idx, 'height'] = int(t2_shape[0])  # Altura
        df_3.loc[idx, 'width'] = int(t2_shape[1])   # Anchura
        df_3.loc[idx, 'depth'] = int(t2_shape[2])   # Profundidad

        # Identificar valores únicos en la máscara de segmentación de lesión
        unique_values = np.unique(processed_images['label'])  

        # Filtrar valores distintos de cero (que representan lesiones con diferentes grados ISUP)
        if len(unique_values) > 1: 
            lesion_values = unique_values[unique_values != 0]
        else:
            lesion_values = unique_values

        if len(lesion_values) > 1:
            print(lesion_values)

        # Registrar el valor ISUP de la lesión de mayor grado (valor más alto)
        if lesion_values.size > 0:
            df_3.loc[idx, 'lesion_value'] = int(lesion_values[-1])
        else:
            df_3.loc[idx, 'lesion_value'] = np.nan

        # Convertir columnas a tipo entero que admite valores nulos
        df_3[['height', 'width', 'depth']] = df_3[['height', 'width', 'depth']].astype(pd.Int32Dtype())
        df_3['lesion_value'] = df_3['lesion_value'].astype(pd.Int32Dtype())
        
    else:
        # Si faltan archivos, asignar NaN a todas las columnas nuevas
        df_3.loc[idx, ['height', 'width', 'depth', 'lesion_value']] = np.nan
```

# Prostate delineations

Esta sección final del código incorpora rutas a las segmentaciones anatómicas de la próstata, completando así el conjunto de datos. Se añaden dos tipos de segmentaciones: 

1. Segmentaciones de la glándula prostática completa
2. Segmentaciones zonales que diferencian la zona periférica (PZ) y la zona de transición (TZ)

Para cada paciente y estudio, el código verifica la existencia de estos archivos de segmentación en formato NIfTI y almacena sus rutas relativas en el DataFrame.

Después de esta adición, el DataFrame contiene toda la información necesaria para análisis posteriores y es guardado como un archivo CSV.

```python
# Definir rutas a los directorios de segmentaciones anatómicas
whole_gland_dir = "../../data/labels/anatomical_delineations/whole_gland/AI/Bosma22b"
zonal_dir = "../../data/labels/anatomical_delineations/zonal_pz_tz/AI/HeviAI23"

# Función para verificar existencia de archivos de segmentación y obtener ruta relativa
def get_path(base_dir, patient_id, study_id):
    # Construir nombre de archivo basado en patient_id y study_id
    file_name = f"{patient_id}_{study_id}.nii.gz"
    file_path = os.path.join(base_dir, file_name)
    
    # Verificar existencia y devolver ruta relativa o None
    if os.path.exists(file_path):
        return file_path.replace("../../", "") # Convertir a ruta relativa
    else:
        return None

# Añadir columna con rutas a segmentaciones de glándula prostática completa
df_3['whole_gland_path'] = df_3.apply(
    lambda row: get_path(whole_gland_dir, row['patient_id'], row['study_id']), axis=1
)

# Añadir columna con rutas a segmentaciones zonales (zona periférica y zona de transición)
df_3['zonal_path'] = df_3.apply(
    lambda row: get_path(zonal_dir, row['patient_id'], row['study_id']), axis=1
)
```

```python
df_3.to_csv("../artifacts/data.csv", index = False)
```
